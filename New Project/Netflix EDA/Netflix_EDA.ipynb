```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
pd.set_option("display.max_columns", None)
```

```python
file_path = Path("netflix_titles.csv")

if not file_path.exists():
    raise FileNotFoundError(
        "Could not find netflix_titles.csv. Download it from Kaggle and place it beside this notebook."
    )

df = pd.read_csv(file_path)
print("Dataset loaded successfully.")
print(f"Rows: {df.shape[0]:,} | Columns: {df.shape[1]:,}")
display(df.head())
```

```python
print("SHAPE:", df.shape)
print("\nDATA TYPES AND NON-NULL COUNTS:")
df.info(memory_usage="deep")

print("\nMEMORY USAGE:")
memory_mb = df.memory_usage(deep=True).sum() / (1024 ** 2)
print(f"{memory_mb:.2f} MB")

print("\nFIRST FIVE ROWS:")
display(df.head())

print("\nLAST FIVE ROWS:")
display(df.tail())

print("\nCOLUMN NAMES:")
print(df.columns.tolist())
```

```python
missing = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percent": (df.isna().mean() * 100).round(2)
}).sort_values("missing_count", ascending=False)

display(missing)

missing_nonzero = missing[missing["missing_count"] > 0]
if missing_nonzero.empty:
    print("There are no missing values in this dataset.")
else:
    ax = missing_nonzero["missing_count"].sort_values().plot(
        kind="barh", figsize=(10, 6), title="Missing Values by Column"
    )
    ax.set_xlabel("Number of missing values")
    ax.set_ylabel("Column")
    plt.tight_layout()
    plt.show()

    top_missing = missing_nonzero.index[0]
    top_count = int(missing_nonzero.loc[top_missing, "missing_count"])
    top_pct = float(missing_nonzero.loc[top_missing, "missing_percent"])
    print(
        f"Interpretation: '{top_missing}' has the most missing values "
        f"({top_count:,}, or {top_pct:.2f}% of rows). "
        "Missing data should be considered before analyses that use this column."
    )
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
if numeric_cols:
    display(df[numeric_cols].describe().T)
    numeric_summary = df[numeric_cols].agg(["mean", "median", "std"]).T
    display(numeric_summary)
else:
    print("No numeric columns were detected.")

categorical_cols = df.select_dtypes(include=["object", "category", "bool"]).columns.tolist()
print("Categorical columns:", categorical_cols)

for col in ["type", "rating"]:
    if col in df.columns:
        print(f"\nValue counts for {col}:")
        display(df[col].value_counts(dropna=False).head(15))
```

```python
analysis = df.copy()

if "date_added" in analysis.columns:
    analysis["date_added"] = pd.to_datetime(
        analysis["date_added"].astype("string").str.strip(),
        errors="coerce"
    )
    analysis["year_added"] = analysis["date_added"].dt.year
    analysis["month_added"] = analysis["date_added"].dt.month

if "duration" in analysis.columns:
    analysis["duration_value"] = pd.to_numeric(
        analysis["duration"].astype("string").str.extract(r"(\d+)")[0],
        errors="coerce"
    )

print("Analysis columns prepared.")
display(analysis.head())
```

```python
if "type" in analysis.columns:
    counts = analysis["type"].value_counts()
    ax = counts.plot(kind="bar", title="Number of Movies vs TV Shows", rot=0)
    ax.set_xlabel("Content type")
    ax.set_ylabel("Number of titles")
    plt.tight_layout()
    plt.show()

    if len(counts) >= 2:
        major = counts.index[0]
        pct = counts.iloc[0] / counts.sum() * 100
        print(f"Interpretation: {major} is the most common content type, representing {pct:.1f}% of non-missing type labels.")
    else:
        print("Interpretation: Only one content type is represented in the available data.")
else:
    print("The required 'type' column is not present.")
```

```python
if "release_year" in analysis.columns:
    yearly = analysis["release_year"].dropna().astype(int).value_counts().sort_index()
    yearly = yearly[yearly.index >= 1990]
    ax = yearly.plot(kind="line", marker="o", title="Titles by Release Year (1990 onward)")
    ax.set_xlabel("Release year")
    ax.set_ylabel("Number of titles")
    plt.tight_layout()
    plt.show()

    if not yearly.empty:
        peak_year = int(yearly.idxmax())
        peak_count = int(yearly.max())
        print(f"Interpretation: Within the displayed period, {peak_year} has the highest title count ({peak_count:,}). This describes the dataset's catalog, not all content produced worldwide.")
else:
    print("The required 'release_year' column is not present.")
```

```python
if "rating" in analysis.columns:
    rating_counts = analysis["rating"].fillna("Missing").value_counts().head(12).sort_values()
    ax = rating_counts.plot(kind="barh", title="Most Common Content Ratings (Top 12)")
    ax.set_xlabel("Number of titles")
    ax.set_ylabel("Rating")
    plt.tight_layout()
    plt.show()

    most_common_rating = analysis["rating"].dropna().mode()
    if not most_common_rating.empty:
        rating = most_common_rating.iloc[0]
        rating_count = int((analysis["rating"] == rating).sum())
        print(f"Interpretation: {rating} is the most frequent recorded rating ({rating_count:,} titles). Ratings describe content classification and should not be interpreted as viewer popularity.")
else:
    print("The required 'rating' column is not present.")
```

```python
if {"type", "duration_value"}.issubset(analysis.columns):
    movies = analysis.loc[
        analysis["type"].eq("Movie") & analysis["duration_value"].notna(),
        "duration_value"
    ]
    if not movies.empty:
        plt.figure(figsize=(10, 6))
        plt.hist(movies, bins=25, edgecolor="black")
        plt.title("Distribution of Movie Durations")
        plt.xlabel("Duration (minutes)")
        plt.ylabel("Number of movies")
        plt.tight_layout()
        plt.show()

        print(
            "Interpretation: The median movie duration is "
            f"{movies.median():.0f} minutes; the middle 50% ranges from "
            f"{movies.quantile(0.25):.0f} to {movies.quantile(0.75):.0f} minutes. "
            "Very short or long titles may form the tails of the distribution."
        )
    else:
        print("No usable movie duration values were found.")
else:
    print("The 'type' or 'duration' column is missing.")
```

```python
if "country" in analysis.columns:
    country_counts = (
        analysis["country"].dropna()
        .str.split(", ")
        .explode()
        .str.strip()
    )
    country_counts = country_counts[country_counts.ne("")].value_counts().head(10).sort_values()

    if not country_counts.empty:
        ax = country_counts.plot(kind="barh", title="Top 10 Countries by Number of Listed Titles")
        ax.set_xlabel("Number of title-country listings")
        ax.set_ylabel("Country")
        plt.tight_layout()
        plt.show()

        top_country = country_counts.idxmax()
        print(
            f"Interpretation: {top_country} has the highest count among the top 10 "
            f"({int(country_counts.max()):,} title-country listings). "
            "Because a title can list multiple countries, these counts can overlap."
        )
    else:
        print("No usable country values were found.")
else:
    print("The required 'country' column is not present.")
```

```python
numeric_for_corr = analysis.select_dtypes(include=np.number)
if numeric_for_corr.shape[1] >= 2:
    corr = numeric_for_corr.corr(numeric_only=True)
    plt.figure(figsize=(8, 6))
    sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
    plt.title("Correlation Heatmap of Numeric Columns")
    plt.tight_layout()
    plt.show()
    print("Interpretation: Read values near +1 as strong positive linear association, values near -1 as strong negative association, and values near 0 as weak linear association. Numeric identifiers, if present, may not have meaningful correlations.")
else:
    print("At least two numeric columns are required for a correlation heatmap.")
```

```python
findings = []

if "type" in analysis.columns and analysis["type"].notna().any():
    type_counts = analysis["type"].value_counts()
    findings.append(
        f"Content mix: {type_counts.index[0]} is the most common type "
        f"({int(type_counts.iloc[0]):,} titles; "
        f"{type_counts.iloc[0] / type_counts.sum() * 100:.1f}% of non-missing type values)."
    )

if "release_year" in analysis.columns:
    years = analysis["release_year"].dropna()
    if not years.empty:
        findings.append(
            f"Release years: the recorded years range from {int(years.min())} "
            f"to {int(years.max())}."
        )

if "rating" in analysis.columns and analysis["rating"].dropna().any():
    mode_rating = analysis["rating"].mode().iloc[0]
    findings.append(
        f"Ratings: {mode_rating} is the most frequently recorded rating "
        f"({int((analysis['rating'] == mode_rating).sum()):,} titles)."
    )

if "country" in analysis.columns:
    cc = analysis["country"].dropna().str.split(", ").explode().str.strip()
    cc = cc[cc.ne("")]
    if not cc.empty:
        findings.append(
            f"Country representation: {cc.value_counts().index[0]} has the most "
            f"title-country listings ({int(cc.value_counts().iloc[0]):,})."
        )

if {"type", "duration_value"}.issubset(analysis.columns):
    movie_durations = analysis.loc[
        analysis["type"].eq("Movie"), "duration_value"
    ].dropna()
    if not movie_durations.empty:
        findings.append(
            f"Movie duration: median duration is {movie_durations.median():.0f} minutes "
            f"across {len(movie_durations):,} movies with usable duration data."
        )

print("KEY FINDINGS")
for i, finding in enumerate(findings[:5], start=1):
    print(f"{i}. {finding}")

print("\nMISSING-DATA NOTE")
print("Review the missing-value table before drawing conclusions. Missing director, cast, or country information may limit analyses that use those fields.")

print("\nYOUR REFLECTION (edit this Markdown cell after running):")
print("- Which result surprised you most, and why?")
print("- What limitations in the dataset could affect your conclusions?")
print("- What additional question would you investigate next?")
```
